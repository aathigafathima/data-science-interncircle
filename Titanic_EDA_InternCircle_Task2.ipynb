```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("titanic")

print(df.head())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df = df.dropna()

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Average age by passenger class:")
print(df.groupby("class")["age"].mean())

print("\nSurvival rate by passenger class:")
print(df.groupby("class")["survived"].mean())
```

```python
print("Key Insights:")
print("1. The dataset was cleaned by handling missing values and duplicate records.")
print("2. Passenger age was analyzed across different passenger classes.")
print("3. Survival rates were compared between passenger classes.")
print("4. Pandas was used for data exploration, cleaning, grouping, and aggregation.")
```
